```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# RMSE values for each model
xgboost_rmse = 0.9959832261071184
random_forest_rmse = 1.0797771831193614
least_squares_rmse = 1.7283776388154193

# Create a DataFrame to store RMSE values for each model
df_rmse = pd.DataFrame({
    'Model': ['XGBoost', 'Random Forest', 'Least Squares'],
    'RMSE': [xgboost_rmse, random_forest_rmse, least_squares_rmse]
})

# Create the plot
plt.figure(figsize=(8, 6))
sns.set_theme(style="whitegrid")

plot = sns.barplot(x='Model', y='RMSE', data=df_rmse, palette='coolwarm', hue='Model')

# Adjust the width of the bars by accessing the patches (bars)
for bar in plot.patches:
    bar.set_width(0.3)  # Adjust the width of the bars here

# Adding plot labels and title
plt.ylabel('RMSE')
plt.title('Comparison of Model Performance')

# Adding the exact RMSE values on top of the bars
for p, rmse_value in zip(plot.patches, [xgboost_rmse, random_forest_rmse, least_squares_rmse]):
    plot.annotate(f'{rmse_value:.6f}', 
                  (p.get_x() + p.get_width() / 2., p.get_height()), 
                  ha='center', va='center', 
                  fontsize=12, color='black', 
                  xytext=(0, 10), textcoords='offset points')

# Show the plot
plt.tight_layout()
plt.show()
```
